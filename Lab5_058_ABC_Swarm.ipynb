```python
import random
random.seed(58) # Setting my own enrolllment random seed as instructed for unique results

def f(x):
    return (x - 7) ** 2 + 4 # The objective function we are trying to minimize

def fitness(x):
    return 1 / (1 + f(x)) # Higher fitness value means we are closer to the true minimum


# Initializing 5 food sources randomly across the continuous search space [0, 14]
sources = [random.uniform(0, 14) for _ in range(5)]

trials = [0 for _ in range(5)]       # Each food source needs its own separate counter to track failed improvements
limit = 8                            # Threshold: if a source doesn't improve after 8 tries, the bee abandons it
iterations = 40                      # Total number of iterations the swarm will run


for it in range(iterations):
    # Employed Bee Phase
    for i in range(5):
        k = random.choice([j for j in range(5) if j != i]) # Pick a different random food source to compare/interact with

        # Move toward or away from source k to explore the local neighborhood
        candidate = sources[i] + random.uniform(-1, 1) * (sources[i] - sources[k])

        # A source only counts as improved if the new spot is strictly better
        if fitness(candidate) > fitness(sources[i]):
            sources[i], trials[i] = candidate, 0      # Keep the new better position and reset the trial counter to 0
        else:
            trials[i] = trials[i] + 1                 # No improvement made, so increment this specific source's failure count

    # Scout Bee Phase
    for i in range(5):
        # If a food source has been stagnant past our limit, abandon it for a fresh random location
        if trials[i] > limit:
            sources[i] = random.uniform(0, 14)        # Abandon stale source and explore somewhere completely new at random
            trials[i] = 0                             # Reset trial counter for the newly discovered scout location

best_x = min(sources, key=f)
print(f"Finished Best x found: {best_x:.4f} | Minimum f(x): {f(best_x):.4f}\n")
```


# **<font color="#0D9488">Experiment 1: Changing the Abandonment Limit (limit = 2** & **limit = 20**)</font>


```python
# EXPERIMENT 1A: Low Limit (limit = 2)
random.seed(58)
limit_low = 2
sources_1 = [random.uniform(0, 14) for _ in range(5)]
trials_1 = [0 for _ in range(5)]

for it in range(40):
    for i in range(5):
        k = random.choice([j for j in range(5) if j != i])
        candidate = sources_1[i] + random.uniform(-1, 1) * (sources_1[i] - sources_1[k])
        if fitness(candidate) > fitness(sources_1[i]):
            sources_1[i], trials_1[i] = candidate, 0
        else:
            trials_1[i] += 1
    for i in range(5):
        if trials_1[i] > limit_low:
            sources_1[i] = random.uniform(0, 14)
            trials_1[i] = 0

print(f"Low Limit (2) Best x: {min(sources_1, key=f):.4f}")


# EXPERIMENT 1B: High Limit (limit = 20)
random.seed(58)
limit_high = 20
sources_2 = [random.uniform(0, 14) for _ in range(5)]
trials_2 = [0 for _ in range(5)]

for it in range(40):
    for i in range(5):
        k = random.choice([j for j in range(5) if j != i])
        candidate = sources_2[i] + random.uniform(-1, 1) * (sources_2[i] - sources_2[k])
        if fitness(candidate) > fitness(sources_2[i]):
            sources_2[i], trials_2[i] = candidate, 0
        else:
            trials_2[i] += 1
    for i in range(5):
        if trials_2[i] > limit_high:
            sources_2[i] = random.uniform(0, 14)
            trials_2[i] = 0

print(f"High Limit (20) Best x: {min(sources_2, key=f):.4f}")
```
